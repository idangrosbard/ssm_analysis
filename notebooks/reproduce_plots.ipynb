# Reproduce Plots

This notebook reproduces the plots from the paper. It first runs the necessary experiments to generate the results, then uses the plot plans defined in `final_plots/plot_plans.json` to generate the final figures.

## 1. Setup and Imports

```python
from src.data_ingestion.data_defs.data_defs import PlotPlans
from src.data_loading.core import load_results_bank
from src.plotting.core import generate_plots_from_plan

# Load the result bank
result_bank = load_results_bank()

# Load plot plans
plot_plans = PlotPlans.load_from_file()

print(f"Loaded {len(plot_plans)} plot plans.")
```

## 2. Generate Plots

```python
for plot_id, plot_plan in plot_plans.items():
    print(f"--- Generating plots for: {plot_plan.title} ---")
    try:
        generate_plots_from_plan(
            plot_plan=plot_plan,
            result_bank=result_bank,
            recreate_plots=True,  # Set to True to regenerate all plots
            save_combined_plot=True,
        )
        print(f"Successfully generated plots for: {plot_plan.title}\n")
    except Exception as e:
        print(f"Error generating plots for {plot_plan.title}: {e}")
```
